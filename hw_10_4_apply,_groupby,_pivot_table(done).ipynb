3.1. Створіть зводну таблицю, яка покаже, скільки покупок (інвойсів) було зроблено в кожній з філій (`Branch`) за різними категоріями товарів. Запишіть таблицю в змінну `invoices_by_category` і виведіть змінну на екран.

3. Є пікові години і дні тижня за прибутком чи виручкою, і ці піки відрізняються по містах чи філіалах.
Що воно нам дає: графік персоналу, можливо промо-акціїї в “тихі” години, інше планування.

```python
def get_hour(time):
    return int(time[:2])

df['Hour'] = df.Time.apply(get_hour)
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['day_name'] = df['Date'].dt.day_name()
```

```python
df.groupby(['day_name', 'Hour'])['gross income'].sum().sort_values(ascending=False).head(10)
```

```python
df.groupby(['day_name', 'Hour'])['gross income'].sum().sort_values(ascending=True).head(10)
```
